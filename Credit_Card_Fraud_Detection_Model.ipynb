```python
# ============================================================
# CREDIT CARD FRAUD DETECTION
# Step 1: Download and Load Dataset
# ============================================================

#pip install -q kagglehub

import kagglehub
import os
import pandas as pd

# Download the official Kaggle Credit Card Fraud Detection dataset
path = kagglehub.dataset_download("mlg-ulb/creditcardfraud")

print("Dataset downloaded successfully!")
print("Dataset location:", path)

# Check downloaded files
print("\nFiles available:")
print(os.listdir(path))

# Load CSV file
df = pd.read_csv(os.path.join(path, "creditcard.csv"))

print("\nDataset loaded successfully!")
print("Dataset shape:", df.shape)

# Display first 5 rows
df.head()
```

```python
# ============================================================
# STEP 2: DATA UNDERSTANDING & DATA QUALITY CHECK
# ============================================================

print("========== DATASET INFORMATION ==========\n")

# Dataset dimensions
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

# Column names
print("\n========== COLUMN NAMES ==========\n")
print(df.columns.tolist())

# Data types
print("\n========== DATA TYPES ==========\n")
print(df.dtypes)

# Missing values
print("\n========== MISSING VALUES ==========\n")
print(df.isnull().sum())

# Duplicate records
print("\n========== DUPLICATE RECORDS ==========\n")
print("Number of duplicate rows:", df.duplicated().sum())

# Statistical summary
print("\n========== STATISTICAL SUMMARY ==========\n")
display(df.describe())

# First 5 records
print("\n========== FIRST 5 RECORDS ==========\n")
display(df.head())
```

```python
# ============================================================
# STEP 3: FRAUD VS GENUINE TRANSACTIONS
# ============================================================

import matplotlib.pyplot as plt
import seaborn as sns

# Count transactions
class_counts = df["Class"].value_counts()

print("Transaction Class Distribution:")
print(class_counts)

print("\nGenuine Transactions (Class 0):", class_counts[0])
print("Fraudulent Transactions (Class 1):", class_counts[1])

# Calculate percentages
genuine_percentage = (class_counts[0] / len(df)) * 100
fraud_percentage = (class_counts[1] / len(df)) * 100

print("\nPercentage of Genuine Transactions:", round(genuine_percentage, 4), "%")
print("Percentage of Fraudulent Transactions:", round(fraud_percentage, 4), "%")

# Visualization
plt.figure(figsize=(8, 5))

sns.countplot(x="Class", data=df)

plt.title("Genuine vs Fraudulent Transactions")
plt.xlabel("Transaction Class")
plt.ylabel("Number of Transactions")

plt.xticks([0, 1], ["Genuine (0)", "Fraud (1)"])

plt.show()
```

```python
# ============================================================
# STEP 4: TRANSACTION AMOUNT ANALYSIS
# ============================================================

plt.figure(figsize=(10, 5))

sns.histplot(
    data=df,
    x="Amount",
    bins=50
)

plt.title("Distribution of Transaction Amount")
plt.xlabel("Transaction Amount")
plt.ylabel("Frequency")

plt.show()
```

```python
# Compare transaction amount for genuine and fraud transactions

plt.figure(figsize=(10, 5))

sns.boxplot(
    x="Class",
    y="Amount",
    data=df
)

plt.title("Transaction Amount: Genuine vs Fraud")
plt.xlabel("Transaction Class")
plt.ylabel("Transaction Amount")

plt.xticks([0, 1], ["Genuine", "Fraud"])

plt.show()
```

```python
# ============================================================
# STEP 5: CORRELATION ANALYSIS
# ============================================================

# Calculate correlation matrix
correlation_matrix = df.corr()

# Display correlation with target variable "Class"
class_correlation = correlation_matrix["Class"].sort_values(ascending=False)

print("Correlation of Features with Fraud (Class):")
print(class_correlation)
```

```python
# ============================================================
# STEP 6: CORRELATION HEATMAP
# ============================================================

plt.figure(figsize=(16, 12))

sns.heatmap(
    correlation_matrix,
    cmap="coolwarm",
    center=0,
    linewidths=0.2
)

plt.title("Correlation Heatmap of Credit Card Transactions")
plt.show()
```

```python
# ============================================================
# STEP 7: TOP FEATURES ASSOCIATED WITH FRAUD
# ============================================================

# Remove Class itself
feature_corr = class_correlation.drop("Class")

# Get strongest correlations
top_positive = feature_corr.abs().sort_values(ascending=False).head(10)

print("Top 10 Features Associated with Fraud:")
print(top_positive)
```

```python
# ============================================================
# STEP 7B: VISUALIZE TOP FEATURES
# ============================================================

plt.figure(figsize=(10, 6))

top_positive.sort_values().plot(kind="barh")

plt.title("Top 10 Features Associated with Fraud")
plt.xlabel("Absolute Correlation")
plt.ylabel("Features")

plt.show()
```

```python
# ============================================================
# STEP 8: FEATURE DISTRIBUTION - FRAUD VS GENUINE
# ============================================================

# Select some important features based on correlation
selected_features = top_positive.index[:6]

print("Selected features:")
print(list(selected_features))

# Plot distributions
for feature in selected_features:

    plt.figure(figsize=(9, 5))

    sns.histplot(
        data=df,
        x=feature,
        hue="Class",
        bins=50,
        stat="density",
        common_norm=False
    )

    plt.title(f"{feature} Distribution: Genuine vs Fraud")
    plt.xlabel(feature)
    plt.ylabel("Density")

    plt.show()
```

```python
# ============================================================
# STEP 9: PREPARE DATA FOR MACHINE LEARNING
# ============================================================

from sklearn.model_selection import train_test_split

# Separate input features and target
X = df.drop("Class", axis=1)
y = df["Class"]

print("Features shape:", X.shape)
print("Target shape:", y.shape)

# Split into training and testing data
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTraining data:")
print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("\nTesting data:")
print("X_test:", X_test.shape)
print("y_test:", y_test.shape)

# Check class distribution
print("\nTraining class distribution:")
print(y_train.value_counts())

print("\nTesting class distribution:")
print(y_test.value_counts())
```

```python
# ============================================================
# STEP 10: FEATURE SCALING
# ============================================================

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

# Make copies to avoid modifying original data
X_train = X_train.copy()
X_test = X_test.copy()

# Fit scaler only on training data
X_train[["Time", "Amount"]] = scaler.fit_transform(
    X_train[["Time", "Amount"]]
)

# Apply same scaler to test data
X_test[["Time", "Amount"]] = scaler.transform(
    X_test[["Time", "Amount"]]
)

print("Feature scaling completed successfully!")

print("\nScaled training data:")
display(X_train.head())
```

```python
# ============================================================
# STEP 11: LOGISTIC REGRESSION MODEL
# ============================================================

from sklearn.linear_model import LogisticRegression

# Create Logistic Regression model
logistic_model = LogisticRegression(
    max_iter=1000,
    class_weight="balanced",
    random_state=42
)

# Train the model
print("Training Logistic Regression model...")

logistic_model.fit(X_train, y_train)

print("Logistic Regression model trained successfully!")

# Make predictions
y_pred_logistic = logistic_model.predict(X_test)

print("Predictions completed!")
```

```python
# ============================================================
# STEP 12: LOGISTIC REGRESSION EVALUATION
# ============================================================

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

# Calculate evaluation metrics
logistic_accuracy = accuracy_score(y_test, y_pred_logistic)
logistic_precision = precision_score(y_test, y_pred_logistic)
logistic_recall = recall_score(y_test, y_pred_logistic)
logistic_f1 = f1_score(y_test, y_pred_logistic)

print("========== LOGISTIC REGRESSION RESULTS ==========")

print("Accuracy :", round(logistic_accuracy, 4))
print("Precision:", round(logistic_precision, 4))
print("Recall   :", round(logistic_recall, 4))
print("F1 Score :", round(logistic_f1, 4))

print("\n========== CLASSIFICATION REPORT ==========")

print(
    classification_report(
        y_test,
        y_pred_logistic,
        target_names=["Genuine", "Fraud"]
    )
)
```

```python
# ============================================================
# STEP 13: LOGISTIC REGRESSION CONFUSION MATRIX
# ============================================================

from sklearn.metrics import ConfusionMatrixDisplay

plt.figure(figsize=(6, 5))

ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_pred_logistic,
    display_labels=["Genuine", "Fraud"],
    cmap="Blues"
)

plt.title("Logistic Regression - Confusion Matrix")
plt.show()
```

```python
# ============================================================
# STEP 14: RANDOM FOREST MODEL
# ============================================================

from sklearn.ensemble import RandomForestClassifier

print("Training Random Forest model...")

random_forest_model = RandomForestClassifier(
    n_estimators=100,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)

# Train Random Forest
random_forest_model.fit(X_train, y_train)

print("Random Forest model trained successfully!")

# Make predictions
y_pred_rf = random_forest_model.predict(X_test)

print("Predictions completed!")
```

```python
# ============================================================
# STEP 15: RANDOM FOREST EVALUATION
# ============================================================

rf_accuracy = accuracy_score(y_test, y_pred_rf)
rf_precision = precision_score(y_test, y_pred_rf)
rf_recall = recall_score(y_test, y_pred_rf)
rf_f1 = f1_score(y_test, y_pred_rf)

print("========== RANDOM FOREST RESULTS ==========")

print("Accuracy :", round(rf_accuracy, 4))
print("Precision:", round(rf_precision, 4))
print("Recall   :", round(rf_recall, 4))
print("F1 Score :", round(rf_f1, 4))

print("\n========== CLASSIFICATION REPORT ==========")

print(
    classification_report(
        y_test,
        y_pred_rf,
        target_names=["Genuine", "Fraud"]
    )
)
```

```python
# ============================================================
# STEP 16: RANDOM FOREST CONFUSION MATRIX
# ============================================================

plt.figure(figsize=(6, 5))

ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_pred_rf,
    display_labels=["Genuine", "Fraud"],
    cmap="Greens"
)

plt.title("Random Forest - Confusion Matrix")
plt.show()
```

```python
# ============================================================
# STEP 17: XGBOOST MODEL
# ============================================================

!pip install -q xgboost

from xgboost import XGBClassifier

print("Training XGBoost model...")

xgb_model = XGBClassifier(
    n_estimators=100,
    max_depth=6,
    learning_rate=0.1,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42,
    eval_metric="logloss",
    n_jobs=-1
)

# Train XGBoost
xgb_model.fit(X_train, y_train)

print("XGBoost model trained successfully!")

# Make predictions
y_pred_xgb = xgb_model.predict(X_test)

print("Predictions completed!")
```

```python
# ============================================================
# STEP 18: XGBOOST EVALUATION
# ============================================================

xgb_accuracy = accuracy_score(y_test, y_pred_xgb)
xgb_precision = precision_score(y_test, y_pred_xgb)
xgb_recall = recall_score(y_test, y_pred_xgb)
xgb_f1 = f1_score(y_test, y_pred_xgb)

print("========== XGBOOST RESULTS ==========")

print("Accuracy :", round(xgb_accuracy, 4))
print("Precision:", round(xgb_precision, 4))
print("Recall   :", round(xgb_recall, 4))
print("F1 Score :", round(xgb_f1, 4))

print("\n========== CLASSIFICATION REPORT ==========")

print(
    classification_report(
        y_test,
        y_pred_xgb,
        target_names=["Genuine", "Fraud"]
    )
)
```

```python
# ============================================================
# STEP 19: XGBOOST CONFUSION MATRIX
# ============================================================

plt.figure(figsize=(6, 5))

ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_pred_xgb,
    display_labels=["Genuine", "Fraud"],
    cmap="Oranges"
)

plt.title("XGBoost - Confusion Matrix")
plt.show()
```

```python
# ============================================================
# STEP 20: FINAL MODEL COMPARISON
# ============================================================

comparison = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "Random Forest",
        "XGBoost"
    ],

    "Accuracy": [
        logistic_accuracy,
        rf_accuracy,
        xgb_accuracy
    ],

    "Precision": [
        logistic_precision,
        rf_precision,
        xgb_precision
    ],

    "Recall": [
        logistic_recall,
        rf_recall,
        xgb_recall
    ],

    "F1 Score": [
        logistic_f1,
        rf_f1,
        xgb_f1
    ]
})

# Round values
comparison = comparison.round(4)

print("========== FINAL MODEL COMPARISON ==========\n")

display(comparison)
```

```python
# ============================================================
# STEP 21: MODEL PERFORMANCE COMPARISON GRAPH
# ============================================================

comparison_plot = comparison.set_index("Model")

comparison_plot.plot(
    kind="bar",
    figsize=(12, 6)
)

plt.title("Performance Comparison of Fraud Detection Models")
plt.xlabel("Machine Learning Model")
plt.ylabel("Score")
plt.ylim(0, 1)

plt.xticks(rotation=0)
plt.legend(title="Evaluation Metrics")

plt.show()
```

```python
# ============================================================
# STEP 22: RANDOM FOREST FEATURE IMPORTANCE
# ============================================================

rf_importance = pd.Series(
    random_forest_model.feature_importances_,
    index=X_train.columns
)

rf_importance = rf_importance.sort_values(
    ascending=False
)

print("Top 15 Important Features - Random Forest\n")
display(rf_importance.head(15))
```

```python
# ============================================================
# STEP 22B: RANDOM FOREST FEATURE IMPORTANCE GRAPH
# ============================================================

plt.figure(figsize=(10, 6))

rf_importance.head(15).sort_values().plot(
    kind="barh"
)

plt.title("Top 15 Feature Importance - Random Forest")
plt.xlabel("Importance")
plt.ylabel("Feature")

plt.show()
```

```python
# ============================================================
# STEP 23: XGBOOST FEATURE IMPORTANCE
# ============================================================

xgb_importance = pd.Series(
    xgb_model.feature_importances_,
    index=X_train.columns
)

xgb_importance = xgb_importance.sort_values(
    ascending=False
)

print("Top 15 Important Features - XGBoost\n")
display(xgb_importance.head(15))
```

```python
# ============================================================
# STEP 23B: XGBOOST FEATURE IMPORTANCE GRAPH
# ============================================================

plt.figure(figsize=(10, 6))

xgb_importance.head(15).sort_values().plot(
    kind="barh"
)

plt.title("Top 15 Feature Importance - XGBoost")
plt.xlabel("Importance")
plt.ylabel("Feature")

plt.show()
```

```python
# ============================================================
# STEP 24: CONFUSION MATRIX VALUES
# ============================================================

cm_logistic = confusion_matrix(y_test, y_pred_logistic)
cm_rf = confusion_matrix(y_test, y_pred_rf)
cm_xgb = confusion_matrix(y_test, y_pred_xgb)

print("Logistic Regression Confusion Matrix:")
print(cm_logistic)

print("\nRandom Forest Confusion Matrix:")
print(cm_rf)

print("\nXGBoost Confusion Matrix:")
print(cm_xgb)
```

```python
# ============================================================
# STEP 25: BEST MODEL SELECTION
# ============================================================

best_model_row = comparison.loc[
    comparison["F1 Score"].idxmax()
]

print("========== BEST MODEL ==========\n")

print("Best Model:", best_model_row["Model"])
print("Accuracy :", best_model_row["Accuracy"])
print("Precision:", best_model_row["Precision"])
print("Recall   :", best_model_row["Recall"])
print("F1 Score :", best_model_row["F1 Score"])
```

```python
# ============================================================
# STEP 26: SAVE MODEL COMPARISON
# ============================================================

comparison.to_csv(
    "fraud_detection_model_comparison.csv",
    index=False
)

print("Model comparison saved successfully!")
```
